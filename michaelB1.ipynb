## Query 2 — Daily Transaction Cost

Return the `trans_date` and the product of `median_transaction_fee` and `transactions` for days where `median_transaction_fee` is greater than 0.5.

The calculated column should be named `daily_cost`.

```python
query2 = bitcoin.loc[
    bitcoin["median_transaction_fee"] > 0.5,
    ["trans_date", "median_transaction_fee", "transactions"]
].copy()

query2["daily_cost"] = (
    query2["median_transaction_fee"] * query2["transactions"]
)

query2 = query2[["trans_date", "daily_cost"]]

query2
```

### Interpretation

The query returned 1,238 observations where the median transaction fee was greater than 0.5. The `daily_cost` was calculated by multiplying the median transaction fee by the number of transactions for each qualifying day. For example, the daily cost on February 27, 2017 was 147,893.720.

as MONEY datatype) and the total amount sent_in_usd (output as MONEY datatype)
for all transactions that occurred in May of 2015. Name the output columns avg_price
and total_transaction_amount, respectively.

1.Write a query that returns each category and the corresponding total revenue for that
category for the sale_year 2016. The aggregated column should be named
total_revenue. The output should be arranged alphabetically.

```python
query1_customer = (
    customer_spending[customer_spending["sale_year"] == 2016]
    .groupby("category", as_index=False)["revenue"]
    .sum()
    .rename(columns={"revenue": "total_revenue"})
    .sort_values("category")
)

query1_customer
```

### Interpretation

In 2016, Bikes generated the highest total revenue among the three categories, with total revenue of 5,722,257. Accessories generated 4,594,897, while Clothing generated 2,079,651. The results show that revenue differed substantially across the product categories during 2016.

2.Write a query that returns a list of sub_categories and their corresponding average
unit_price (named avg_unit_price), average unit_cost named avg_unit_cost, as well as
the difference between these two values (named margin) for the sale_year 2015. Round
all values to two decimal places. Organize the results alphabetically.

```python
query2_customer = (
    customer_spending[customer_spending["sale_year"] == 2015]
    .groupby("sub_category")
    .agg(
        avg_unit_price=("unit_price", "mean"),
        avg_unit_cost=("unit_cost", "mean")
    )
    .reset_index()
)

query2_customer["margin"] = (
    query2_customer["avg_unit_price"] -
    query2_customer["avg_unit_cost"]
)

query2_customer[
    ["avg_unit_price", "avg_unit_cost", "margin"]
] = query2_customer[
    ["avg_unit_price", "avg_unit_cost", "margin"]
].round(2)

query2_customer = query2_customer.sort_values("sub_category")

query2_customer
```

4.Write a query that returns the age, sub_cateogry, average quantity (as a whole
number, named avg_quantity), and average cost of products (rounded to 2 decimals,
named avg_cost) purchased by each age and sub_category. Organize the data by age,
oldest to youngest, and then by sub_category alphabetically.

```python
query4_customer = (
    customer_spending
    .groupby(["age", "sub_category"])
    .agg(
        avg_quantity=("quantity", "mean"),
        avg_cost=("cost", "mean")
    )
    .reset_index()
)

query4_customer["avg_quantity"] = (
    query4_customer["avg_quantity"].round().astype(int)
)

query4_customer["avg_cost"] = (
    query4_customer["avg_cost"].round(2)
)

query4_customer = query4_customer.sort_values(
    ["age", "sub_category"],
    ascending=[False, True]
)

query4_customer
```

### Interpretation

The results show how average purchase quantity and average product cost vary across customer ages and sub-categories. The dataset includes customers ranging from age 17 to age 87. For each age, the results are further divided by sub-category, allowing differences in purchasing quantity and product cost to be examined across both customer age and product type.

6.Write a query to identify which sale_year and category combinations were the most
profitable. Return the columns sale_year, category, the total revenue as total_revenue,
the total cost as total_cost, and the difference between the two as profit for each
sale_year and category. Include only rows that were profitable and sort the results from
highest profit to lowest.

```python
query6_customer = (
    customer_spending
    .groupby(["sale_year", "category"])
    .agg(
        total_revenue=("revenue", "sum"),
        total_cost=("cost", "sum")
    )
    .reset_index()
)

query6_customer["profit"] = (
    query6_customer["total_revenue"] -
    query6_customer["total_cost"]
)

query6_customer = query6_customer[
    query6_customer["profit"] > 0
]

query6_customer = query6_customer.sort_values(
    "profit",
    ascending=False
)

query6_customer
```

### Interpretation

The results show that five sale year and category combinations were profitable. The highest profit was generated by Accessories in 2016, with a profit of 1,035,251. This was followed by Bikes in 2016, with a profit of 512,935, and Clothing in 2016, with a profit of 424,796. Accessories in 2015 generated a profit of 343,518, while Clothing in 2015 generated 120,436.

Only profitable combinations are shown because combinations with zero or negative profit were excluded according to the query requirement.

7.Write a query that calculates the average spending per age for male customers
(gender), where spending is based on the product of unit_price and quantity. Return the age and the calculated average spending, rounded to two decimal places, as
avg_spending. Organize the results from highest to lowest avg_spending.

```python
query7_customer = (
    customer_spending[
        customer_spending["gender"] == "M"
    ]
    .assign(
        spending=lambda x: x["unit_price"] * x["quantity"]
    )
    .groupby("age")
    .agg(
        avg_spending=("spending", "mean")
    )
    .reset_index()
)

query7_customer["avg_spending"] = (
    query7_customer["avg_spending"].round(2)
)

query7_customer = query7_customer.sort_values(
    "avg_spending",
    ascending=False
)

query7_customer
```

### Interpretation

The results show that average spending varies substantially across ages among male customers. The highest average spending is observed for customers aged 86, at 2,026.00. Other relatively high averages occur among customers aged 73, 74, and 75. Because the results are ordered from highest to lowest average spending, the table makes it easy to identify the age groups with the largest average spending values.

8.Write a query to determine the highest unit_cost, lowest unit_cost, and average
unit_cost for each gender in each category. The output columns should be gender,
category, high_cost, low_cost, avg_cost, in that order. Organize the results by gender
and then category.

```python
query8_customer = (
    customer_spending
    .groupby(["gender", "category"])
    .agg(
        high_cost=("unit_cost", "max"),
        low_cost=("unit_cost", "min"),
        avg_cost=("unit_cost", "mean")
    )
    .reset_index()
)

query8_customer["high_cost"] = query8_customer["high_cost"].round(2)
query8_customer["low_cost"] = query8_customer["low_cost"].round(2)
query8_customer["avg_cost"] = query8_customer["avg_cost"].round(2)

query8_customer = query8_customer[
    ["gender", "category", "high_cost", "low_cost", "avg_cost"]
].sort_values(
    ["gender", "category"]
)

query8_customer
```

### Interpretation

The results show differences in unit costs across product categories and gender groups. Bikes have the highest average unit cost for both female and male customers, at 956.65 and 949.31 respectively. Accessories have much lower average unit costs, while Clothing falls between the two categories. The highest observed unit cost is 3,240.00 for male customers in Accessories, compared with 3,120.00 for female customers in the same category.

9.Write a query to determine the age distribution of customers by category and country
for the sale_year 2016. For each category and country, calculate the age of the
youngest and oldest customers, and the average age of customers rounded to one
decimal place. The output columns should be category, country, youngest_customer,
oldest_customer, and avg_customer_age, in that order. Organize your results by
category and then by the average age of customers.

```python
query9_customer = (
    customer_spending[
        customer_spending["sale_year"] == 2016
    ]
    .groupby(["category", "country"])
    .agg(
        youngest_customer=("age", "min"),
        oldest_customer=("age", "max"),
        avg_customer_age=("age", "mean")
    )
    .reset_index()
)

query9_customer["avg_customer_age"] = (
    query9_customer["avg_customer_age"].round(1)
)

query9_customer = query9_customer.sort_values(
    ["category", "avg_customer_age"]
)

query9_customer
```

### Interpretation

The 2016 results show differences in customer age distributions across product categories and countries. The youngest customer in every category-country combination was age 17. Average customer age varied across the combinations, with Bikes in the United States having the highest average age at 39.7 years. The results also show that the United States generally had higher average customer ages than the other countries within each product category.

10.Write a query to return the country that has the highest average revenue (rounded to 2
decimals). Your output columns should be country and high_sales.

```python
query10_customer = (
    customer_spending
    .groupby("country")
    .agg(
        high_sales=("revenue", "mean")
    )
    .reset_index()
)

query10_customer["high_sales"] = (
    query10_customer["high_sales"].round(2)
)

query10_customer = query10_customer.sort_values(
    "high_sales",
    ascending=False
).head(1)

query10_customer
```

### Interpretation

Germany has the highest average revenue among the countries in the dataset, with an average revenue of 816.09 per transaction record. This result identifies Germany as the country with the highest mean revenue based on the available customer spending data.
