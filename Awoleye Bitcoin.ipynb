Write a query that returns the trans_date along with the code_size divided by the 
transactions. Show all decimals that would normally result from division and name the 
calculated column difficulty. 

```python
# The trans_date along with the code_size divided by the transactions, Called difficulty.
data1 = data[['trans_date']].copy()
data1['difficulty'] = data['code_size'] / data['transactions']
data1
```

Write a query that returns the trans_date and the product of median_transaction_fee 
and transactions for days where the median_transaction_fee is more than 0.5. Label 
the calculated column daily_cost.

```python
# The trans_date and the product of median_transaction_fee and transactions, called daily_cost. 
# Filter the data to only include rows where median_transaction_fee is greater than 0.5.
filtered_data = data[data['median_transaction_fee'] > 0.5].copy()
filtered_data['daily_cost'] = filtered_data['median_transaction_fee'] * filtered_data['transactions']
filtered_data[['trans_date', 'daily_cost']]
```

Write a query that returns the trans_date and the sent_in_usd divided by the number of 
transactions on days where there were more than 400,000 transactions. Show all 
expected decimals. Name the output column avg_transaction.

```python
# The trans_date and the average transaction value, which is the sent_in_usd divided by the transactions, called avg_transaction.
filtered_transactions = data[data['transactions'] > 400000].copy()
filtered_transactions['avg_transaction'] = filtered_transactions['sent_in_usd'] / filtered_transactions['transactions']
filtered_transactions[['trans_date', 'avg_transaction']]
```

Write a query that returns the average price_usd (rounded to 2 decimal places, output 
as MONEY datatype) and the total amount sent_in_usd (output as MONEY datatype) 
for all transactions that occurred in May of 2015. Name the output columns avg_price 
and total_transaction_amount, respectively.

```python
# Average price and total amount sent for transactions in May 2015.
may_2015 = data[pd.to_datetime(data['trans_date']).dt.to_period('M') == '2015-05']
may_2015_summary = pd.DataFrame({
    'avg_price': [may_2015['price_usd'].mean().round(2)],
    'total_transaction_amount': [may_2015['sent_in_usd'].sum()]
})
may_2015_summary.assign(
    avg_price=may_2015_summary['avg_price'].map(lambda value: f'${value:,.2f}'),
    total_transaction_amount=may_2015_summary['total_transaction_amount'].map(lambda value: f'${value:,.2f}')
)
```

Write a query that returns the smallest and largest market_cap values as whole 
numbers and the lowest and highest price_usd rounded to 2 decimals. Name these 
columns as low_cap, high_cap, low_price_usd, and high_price_usd respectively.

```python
# Smallest and largest market_cap values and lowest and highest price_usd values.
market_price_summary = pd.DataFrame({
    'low_cap': [int(data['market_cap'].min())],
    'high_cap': [int(data['market_cap'].max())],
    'low_price_usd': [round(data['price_usd'].min(), 2)],
    'high_price_usd': [round(data['price_usd'].max(), 2)]
})
market_price_summary
```

Write a query that finds the lowest median_transaction_fee, the highest 
median_transaction_fee, and the average transaction_fees for trans_dates between 
August 10, 2017 and August 10, 2019 (inclusive). Round the average to 5 decimal 
places. Name these columns lowest_fee, highest_fee, and avg_fee respectively.

```python
# Fee summary for trans_dates from August 10, 2017 through August 10, 2019, inclusive.
data['trans_date'] = pd.to_datetime(data['trans_date'])
fee_period = data[
    (data['trans_date'] >= '2017-08-10') &
    (data['trans_date'] <= '2019-08-10')
]
fee_summary = pd.DataFrame({
    'lowest_fee': [fee_period['median_transaction_fee'].min()],
    'highest_fee': [fee_period['median_transaction_fee'].max()],
    'avg_fee': [round(fee_period['transaction_fees'].mean(), 5)]
})
fee_summary
```

Write a query that shows the average google_trends score (as a whole number) and

Write a query that returns the minimum, maximum, and average confirmation_time for 
all transactions. Round the average to three decimal places. Name the output columns 
min_confirmation_time, max_confirmation_time, and avg_confirmation_time

Write a query that returns the averages for the following columns (all rounded to 2 
decimals): price_usd, mining_profitability, and transaction_fees for all transactions 
that happened between March 1, 2020, and November 30, 2020. Name the columns: 
avg_price_usd, avg_mining_profitability, and avg_transaction_fees respectively.

```python
# Average values from March 1, 2020 through November 30, 2020, inclusive.
transaction_dates = pd.to_datetime(data['trans_date'])
period_2020 = data[
    (transaction_dates >= '2020-03-01') &
    (transaction_dates <= '2020-11-30')
]
average_summary_2020 = pd.DataFrame({
    'avg_price_usd': [round(period_2020['price_usd'].mean(), 2)],
    'avg_mining_profitability': [round(period_2020['mining_profitability'].mean(), 2)],
    'avg_transaction_fees': [round(period_2020['transaction_fees'].mean(), 2)]
})
average_summary_2020
```
